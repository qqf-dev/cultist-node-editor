```python
import os
import json

IMAGES_DIR = './origin_resources/images'
OUTPUT_FILE = 'images-manifest.json'
IMAGE_EXTS = ('.png', '.jpg', '.jpeg', '.gif', '.webp', '.PNG')

manifest = {}

for root, dirs, files in os.walk(IMAGES_DIR):
    # 计算相对路径作为分类 key
    rel_dir = os.path.relpath(root, IMAGES_DIR).replace('\\', '/')
    if rel_dir == '.':
        rel_dir = ''
    for file in files:
        if file.lower().endswith(IMAGE_EXTS):
            if rel_dir not in manifest:
                manifest[rel_dir] = []
            manifest[rel_dir].append(file)

# 排序
for key in manifest:
    manifest[key].sort()

with open(OUTPUT_FILE, 'w', encoding='utf-8') as f:
    json.dump(manifest, f, indent=2)

print(f'✅ manifest.json 已生成至 {OUTPUT_FILE}')
```
